```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("allanbruno/airbnb-rio-de-janeiro")

print("Path to dataset files:", path)
```

```python
import os
print (os.listdir(path))
```

```python
import pandas as pd
import os

arquivo = os.path.join(path, "total_data.csv")

dados = pd.read_csv(arquivo)

print(dados.shape)
print(dados.head())
```

```python
print(dados.columns.tolist())
```

```python
dados.info()
```

```python
colunas = [
    'id',
    'neighbourhood_cleansed',
    'room_type',
    'property_type',
    'accommodates',
    'bedrooms',
    'price',
    'number_of_reviews',
    'review_scores_rating',
    'latitude',
    'longitude',
    'month'
]

dados_analise = dados[colunas]

print(dados_analise.shape)
print(dados_analise.head())
```

```python
print(dados_analise['price'].dtype)

print(dados_analise['price'].head(10))
```

```python
# Remover o símbolo de dólar e as vírgulas
dados_analise['price'] = (
    dados_analise['price']
    .str.replace('$', '', regex=False)
    .str.replace(',', '', regex=False)
)

# Converter os preços para números
dados_analise['price'] = pd.to_numeric(
    dados_analise['price'],
    errors='coerce'
)

# Conferir o resultado
print(dados_analise['price'].dtype)
print(dados_analise['price'].head(10))
```

```python
print(dados_analise['price'].describe())
```

```python
# Criar uma cópia independente das colunas selecionadas
dados_analise = dados[colunas].copy()

# Limpar e converter a coluna de preços
dados_analise.loc[:, 'price'] = pd.to_numeric(
    dados_analise['price']
    .str.replace('$', '', regex=False)
    .str.replace(',', '', regex=False),
    errors='coerce'
)

print(dados_analise['price'].describe())
```

```python
# Quantidade de preços ausentes
print("Preços ausentes:", dados_analise['price'].isna().sum())

# Quantidade de preços iguais a zero
print("Preços iguais a zero:", (dados_analise['price'] == 0).sum())

# Quantidade de preços negativos
print("Preços negativos:", (dados_analise['price'] < 0).sum())

# Dez maiores preços
print(
    dados_analise['price']
    .nlargest(10)
    .to_list()
)
```

```python
# Criar uma cópia independente
dados_analise = dados[colunas].copy()

# Converter os preços para números
dados_analise['price'] = pd.to_numeric(
    dados_analise['price']
    .astype('string')
    .str.replace('$', '', regex=False)
    .str.replace(',', '', regex=False),
    errors='coerce'
)

# Conferir o tipo da coluna
print("Tipo de price:", dados_analise['price'].dtype)

# Investigar os valores
print("Preços ausentes:", dados_analise['price'].isna().sum())
print("Preços iguais a zero:", (dados_analise['price'] == 0).sum())
print("Preços negativos:", (dados_analise['price'] < 0).sum())

# Mostrar os dez maiores preços
print("Dez maiores preços:")
print(dados_analise['price'].nlargest(10).to_list())
```

```python
# Selecionar os 10 anúncios mais caros
mais_caros = dados_analise.nlargest(10, 'price')

# Exibir as informações disponíveis
colunas_exibicao = [
    'id',
    'neighbourhood_cleansed',
    'room_type',
    'property_type',
    'accommodates',
    'price'
]

print(mais_caros[colunas_exibicao].to_string(index=False))
```

```python
# Selecionar os anúncios com preço zero
precos_zero = dados_analise[
    dados_analise['price'] == 0
].copy()

# Recuperar os nomes na base original
precos_zero['name'] = dados.loc[
    precos_zero.index, 'name'
]

# Exibir os primeiros 10 anúncios
print("Total de anúncios com preço zero:", len(precos_zero))

print(
    precos_zero[
        ['id', 'name', 'neighbourhood_cleansed', 'room_type', 'price']
    ].head(10).to_string(index=False)
)
```
